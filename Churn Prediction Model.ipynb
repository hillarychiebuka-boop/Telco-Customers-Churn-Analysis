```python
from imblearn.combine import SMOTEENN

sm = SMOTEENN(random_state = 100)
x_resampled, y_resampled = sm.fit_resample(x, y)
```

In churn prediction, usually fewer customers churn compared to those who stay.

Without balancing, the model might ignore churners.

- SMOTEENN helps the model pay equal attention to both churners and non-churners, improving recall (catching actual churners).

* SMOTE (Synthetic Minority Oversampling Technique) → creates “new bananas” by slightly modifying existing ones (like making copies with small changes).

* ENN (Edited Nearest Neighbors) → cleans out noisy or confusing apples that don’t fit well.

**sm = SMOTEENN():** You are setting up a "Data Fixer" to handle the fact that you have way more "Stay" customers than "Churn" customers.

**sm.fit_resample(x, y):**

The "SMOTE" part: It looks at the customers who left and creates "synthetic" (fake but realistic) profiles of more customers who left. This gives the model more examples of "leaving behavior" to study.

The "ENN" part: It deletes the "noisy" data points—like a customer who looks exactly like someone who would stay, but left anyway for a random reason. This makes the patterns of "churning" much clearer for the computer.

The Result: x_resampled and y_resampled now contain a balanced mix of "Stayers" and "Leavers," so your model can't just take the easy way out by guessing "Stay" every time.


```python
xr_train, xr_test, yr_train,yr_test = train_test_split(x_resampled, y_resampled, test_size = 0.2)
```


This code is doing exactly what was done earlier with train_test_split, but now it's applied to the balanced dataset after SMOTEENN resampling.

train_test_split(...)

```python
print(metrics.confusion_matrix(yr_test, yr_predict))
```

metrics.confusion_matrix(...) → Creates a table that compares the true labels (yr_test) with the predicted labels (yr_predict).

The output is a 2×2 matrix (for binary classification like churn vs. non-churn):

[[TN, FP]

[FN, TP]]

Where:

TN (True Negative) → Correctly predicted “no churn.”

FP (False Positive) → Predicted “churn” when it was actually “no churn.”

FN (False Negative) → Predicted “no churn” when it was actually “churn.”

TP (True Positive) → Correctly predicted “churn.”

* The confusion matrix is basically the scorecard showing where the child got it right and where they got confused.

Accuracy alone doesn’t tell the whole story.

The confusion matrix shows exactly where the model struggles:

If FN is high → the model is missing churners (bad for business).

If FP is high → the model is falsely flagging loyal customers as churners.

With SMOTEENN balancing, you should see better recall (fewer missed churners).

**Performing PCA** (Principal Component Analysis)


```python
# Applying PCA
from sklearn.decomposition import PCA
pca = PCA(0.9)
xr_train_pca = pca.fit_transform(xr_train1)
xr_test_pca = pca.transform(xr_test1)
explained_variance = pca.explained_variance_ratio_
```

Building on our "Super Column" analogy, this code is telling the computer: "Shrink my data, but don't lose the important stuff."

Here is the breakdown of what each line is doing to your churn data:

from sklearn.decomposition import PCA: You are reaching into the Scikit-Learn toolkit and grabbing the PCA "Compressor" tool.
pca = PCA(0.9): This is the most important part. By putting 0.9 inside the brackets, you are giving a specific instruction: "Create enough 'Super Columns' to keep 90% of the original information (variance), and throw away the 10% that is just noisy junk."

xr_train_pca = pca.fit_transform(xr_train1):
**Fit:** The computer studies your training data to find the best way to combine your 50+ columns into "Super Columns."
Transform: It then actually "squashes" the data down into those new columns.

xr_test_pca = pca.transform(xr_test1): You apply the exact same squash to your test data. Notice there is no "fit" here—you want to treat the test data exactly how you treated the training data so the "exam" is fair.

explained_variance = pca.explained_variance_ratio_: This creates a list showing how much "power" each new Super Column has. For example, if the first number is 0.70, it means Super Column 1 alone holds 70% of all the information about why customers stay or leave.

The Essence for your Churn Model:
Before this code, your xr_train1 might have had 50 columns. After this code, xr_train_pca might only have 10 or 15 columns, but those columns are so "dense" with information that your model will run faster and be less likely to get "distracted" by minor, irrelevant details.

```python
model = RandomForestClassifier(n_estimators = 100, criterion = 'gini', random_state = 100, max_depth = 6, min_samples_leaf = 8)
model.fit(xr_train_pca, yr_train1)
```

This code is like hiring a "Committee of 100 Experts" (Random Forest) and giving them very specific rules on how to judge whether a customer will churn.
Since you are using xr_train_pca, you are giving these experts the "Super Columns" we talked about earlier instead of the raw, messy data.

Line-by-Line Breakdown:
n_estimators = 100: This tells the computer to build 100 Decision Trees. Each tree will look at the data slightly differently, and they will vote at the end to give you a final answer.

criterion = 'gini': This is the "Math Rule" the trees use to decide how to split the data. Think of it as a "Purity Test." The tree asks questions that try to separate "Churners" from "Stayers" as cleanly as possible.

random_state = 100: This is your "Save Slot." It ensures that if you run this code again, the 100 trees are built in the exact same way so your results don't change.

max_depth = 6: This is a "Safety Limit." It tells each tree, "You can only ask 6 questions before you have to make a decision." This stops the trees from becoming too complex and "over-memorizing" the data.

min_samples_leaf = 8: This means a "branch" on the tree must represent at least 8 customers to be valid. If a pattern only applies to 1 or 2 weird customers, the tree is told to ignore it. This keeps the model focused on general trends rather than individual outliers.

model.fit(xr_train_pca, yr_train1): This is the "Study Session." The 100 trees look at the "Super Columns" (xr_train_pca) and the correct answers (yr_train1) to learn the patterns of churn.

Why these specific settings?

By setting max_depth and min_samples_leaf, you are intentionally making the trees "simpler." This is a smart move in Churn Prediction because it helps the model work well on new, future customers rather than just being perfect on the old data you already have.

```python
yr_predict_pca = model.predict(xr_test_pca)
model_score_r_pca = model.score(xr_test_pca, yr_test1)
```

Building on our previous analogies, this code is the "Final Exam" for your committee of 100 experts.

Since you trained them using the "Super Columns" (PCA), you are now testing them to see if they can accurately predict churn on data they have never seen before.

**Line-by-Line Breakdown:**

yr_predict_pca = model.predict(xr_test_pca):

The Action: You are handing the "Super Columns" of your test group (xr_test_pca) to your Random Forest.

Layman's terms: This is the model actually taking the test. It looks at the simplified patterns it learned and writes down its "Guess" (Churn or Stay) for every single customer in that list.

The Result: All those guesses are saved in a new list called yr_predict_pca.

model_score_r_pca = model.score(xr_test_pca, yr_test1):
The Action: This is a quick way to get the Accuracy Percentage.

Layman's terms: The computer looks at the model's guesses, compares them to the "Answer Key" (yr_test1), and gives you a single number (like 0.78).

The Result: If it shows 0.78, it means your model got 78% of the questions right.

Why is there a "_pca" at the end of the names?

In coding, this is a best practice. By adding _pca to your variables, you are keeping this model's results separate from your previous model (the one that used all the messy, original columns). This allows you to compare them side-by-side.

The Moment of Truth:

If your model_score_r_pca is higher than your original score, it means the PCA "Super Columns" successfully cleared away the noise and helped the model see the truth more clearly.

```python
print(model_score_r_pca)
print(metrics.classification_report(yr_test1, yr_predict_pca))
```

**Difference between Decision Tree & Random Forest**

Think of these two models as a Manager versus a Committee.

1. **Decision Tree (The Single Manager):**
Imagine you have one manager deciding if a customer will churn (leave). They use a flow chart of "Yes/No" questions:

"Has the customer been with us for less than 6 months?" → Yes

"Do they have a high monthly bill?" → Yes

"Have they called support 3 times this month?" → Yes

Result: Prediction = Churn.

**The Downside:** A single manager might be biased or "overthink" a specific customer's weird habits (this is called Overfitting). If that one manager is wrong, the whole prediction is wrong.

2. **Random Forest (The Committee):** A Random Forest is just a large group (a "forest") of many Decision Trees working together.

**How it works:**
The Crowd: Instead of one manager, you hire 100 managers.

Different Perspectives: You don't give every manager the same data. You give each one a slightly different "shuffled" version of the customer list and only show them a few specific details (like only their "Contract Type" or only their "Internet Speed").

The Vote: Each tree makes its own independent "Yes/No" prediction.

The Final Answer: You take a vote. If 80 trees say "Churn" and 20 say "Stay," the Random Forest predicts Churn.

**Why is the Forest better?**

"The Wisdom of the Crowd."

Even if a few trees get confused by a weird data point, the majority of the trees will usually pull the answer back to the truth. This makes the Random Forest much more stable and accurate than a single Decision Tree.

In the project:
Decision Tree: Might focus too hard on one specific rule.
Random Forest: Will look at the "big picture" of your customer data

**Principal Component Analysis**

PCA stands for Principal Component Analysis — it’s a technique in machine learning and statistics used for dimensionality reduction.

In short: PCA is a way to compress your dataset into fewer dimensions while keeping most of the important information. It’s especially useful when you have many features, like in your churn prediction project.

**What PCA does, it:**

Takes a dataset with many features (columns).

Finds new “directions” (called principal components) that capture the most variance (information) in the data.

Projects the data onto these components, reducing the number of dimensions while keeping most of the important information.

**Interpretation of each Model_Score**

This is the most important "Aha!" moment in your learning so far. At first glance, it looks like the Decision Tree (92%) is the winner, but in the world of Churn Prediction, it is likely the biggest liar.

Here is how to interpret these results in layman's terms:

1. The Decision Tree (92%) – "The Student who Memorized the Practice Test"
In machine learning, we call this Overfitting.

What happened: A single Decision Tree is very "bossy." Because you didn't give it strict rules (like max_depth), it likely memorized every single specific detail of your training data—even the random coincidences.

The Churn Effect: If a customer named "John" left and he happened to wear a "Red Shirt," the tree might make a rule: "Everyone in a Red Shirt will leave." This works perfectly on your current data, but it will fail miserably on new, real-world customers who don't fit that exact pattern.

2. The Random Forest (78%) – "The Wise Committee".
This is likely your most reliable model.

What happened: By using 100 trees and setting rules like max_depth=6, you forced the model to look at general trends instead of memorizing individuals. It dropped from 92% to 78% because it stopped "cheating" by memorizing and started actually "learning."

The Churn Effect: This model is more "robust." It will probably perform much better on next month's customer data than the Decision Tree will.

1. The PCA Model (70%) – "The Simplified View"
What happened: By using PCA, you threw away 10% of your data's information to make it simpler. In this case, you might have thrown away a few details that were actually important for predicting churn.

The Churn Effect: While 70% is lower, it tells us that even with a "blurry" version of the data, the model can still get 7 out of 10 customers right.

**The Verdict for your Project:**

In a Churn Project, High Accuracy is often a Trap.
If 90% of your customers stay, a "dumb" model can get 90% accuracy just by guessing "Stay" every time—but it will miss 100% of the people who actually leave.

To find the real winner, you must look at the "Recall" for the "Churn" class:
Check your classification_report for each model.

Look for the Recall score for Class 1 (the Churners).
The model with the highest Recall for Churn is your true winner, because that’s the one that actually "caught" the most people before they left.

**Customer churn prediction overview**

Customer churn prediction predicts the likelihood of customers canceling a company’s products or services. In most cases, businesses with repeat clients or clients under subscriptions strive to maintain the customer base. Therefore, it is important to keep track of the customers who cancel their subscription plan and those who continue with the service. This approach requires the organization to know and understand their client’s behavior and the attributes that lead to the risk of the client leaving. I will explain the steps necessary in creating and deploying a churn prediction machine-learning model.

Why Predict Customer Churn?

It is important for any organization dealing with repeat clients to find ways to retain existing ones. The approach is crucial since customer churn is expensive, and acquiring new clients is more expensive than retaining existing ones. Consider an internet service (ISP) provider who has acquired a new user. They will need technicians and hardware to connect the latest client to their service. The client will only be required to pay the subscription fee to continue using their plan. If the user fails to renew their service, the company will most likely be at a loss, especially if the trend continues for several customers. The monthly recurring revenue (MRR) for such an institution will likely be low; hence, it will be unable to sustain the business. Thus, a reliable churn prediction model should help companies stay afloat as they scale up and attract more customers.

Case Study of Customer Churn Prediction Model

Creating churn prediction models involves using historical customer data to predict the likelihood of the current customer leaving or continuing with a particular service/product. The data used for the predictive models include product usage data and direct customer feedback. Besides, the predictive models identify the different trends and patterns in the data to forecast customer churn.

Scenario

Consider an e-commerce company with historical data on how its clients have interacted with its services. The company wants to know the likelihood of customers churning so it can launch targeted marketing campaigns.
